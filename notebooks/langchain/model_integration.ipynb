Model Integration With Google Gemini and Groq

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GEMINI_API_KEY"] = os.getenv("GEMINI_API_KEY")
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model = init_chat_model("google_genai:gemini-3.5-flash")
model
```

```python
response = model.invoke("hello how are you?")
response
```

We can also do with ChatGroq lib , ChatOpenai lib

Streaming (we able to see result in chunks as ai is generating we dont have to wait for complete execution) and Batches (giving multiple inputs parallely)

```python
model.invoke("write me a 200 words paragarph on ai")
```

```python
for chunk in model.stream("write me a 200 words paragarph on ai"):
    print(chunk.text,end="|",flush=True)
```

```python
responses = model.batch([
    "why parrots are green",
    "why do airplane fly",
    "what is computing"
])
for response in responses:
    print(response)
```

```python
responses = model.batch([
    "why parrots are green",
    "why do airplane fly",
    "what is computing"
],
    config={
        'max_concurrency': 5, # max parallel calls
    }
)
for response in responses:
    print(response.text)
```
