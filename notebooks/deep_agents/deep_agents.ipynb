```python
import os
from typing import Literal
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["Tavily_API_KEY"] = os.getenv("Tavily_API_KEY")
```

```python
from tavily import TavilyClient
tavily_client = TavilyClient(api_key=os.getenv("Tavily_API_KEY"))

def web_search(query:str,max_results:int=5,topic:Literal["general","sports","news","finance"]="general",include_raw=False):
    """Run a WebSearch"""
    return tavily_client.search(query=query,max_results=max_results,include_raw=include_raw,topic=topic)
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)
```

```python
## Creating deep agent
from deepagents import create_deep_agent
deep_agent = create_deep_agent(
    model=llm,
    tools=[web_search],
    system_prompt="Act as researcher"
)
deep_agent
```

```python
result = deep_agent.invoke({"messages":[{"role":"user","content":"What is cyst in liver"}]})
result
```
