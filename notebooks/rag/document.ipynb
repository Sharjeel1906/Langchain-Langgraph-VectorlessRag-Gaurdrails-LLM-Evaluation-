```python
### Create a simple text file
import os
os.makedirs("./data/text_files",exist_ok=True)
```

```python
sample_texts = {
    "../data/text_files/python_intro.txt": """Python Programming Introduction

Python is a high-level, interpreted, general-purpose programming language known for its simplicity, readability, and versatile ecosystem. Created by Guido van Rossum and first released in 1991, Python's design philosophy emphasizes code readability with its notable use of significant whitespace. Over the decades, it has evolved into one of the most widely used programming languages across software engineering, web development, data science, machine learning, and automation.

Key Features & Paradigms:
- Simple and Readable Syntax: Clean language construct that resembles natural English, reducing the cost of program maintenance.
- Dynamic Typing & Automatic Memory Management: Enables fast prototyping without manual memory allocation or explicit variable declarations.
- Extensive Standard Library: Often described as having "batteries included," providing built-in modules for string handling, file I/O, networking, and protocols.
- Object-Oriented, Functional, and Imperative: Flexible multiparadigm programming support to structure code logically based on project demands.
- Cross-Platform Compatibility: Seamlessly runs across Linux, macOS, and Windows operating systems without requiring core code changes.
- Massive Open-Source Ecosystem: Supported by PyPI (Python Package Index), housing hundreds of thousands of third-party libraries.

Common Applications:
1. Web & Backend Development: Powering scalable web services using robust frameworks like Django, FastAPI, and Flask.
2. Data Science & Analytics: Manipulating, processing, and analyzing large datasets with NumPy, Pandas, and SciPy.
3. Machine Learning & AI: Building and training modern neural networks using PyTorch, TensorFlow, and Scikit-learn.
4. Scripting & Task Automation: Writing lightweight scripts to handle repetitive system administration tasks, web scraping, and automated workflows.
5. Desktop & Embedded Systems: Constructing cross-platform GUI applications using Tkinter, PyQt, or running micropython on microcontrollers.

Python continues to be an essential tool for beginners learning computer science principles as well as experienced engineers deploying large-scale production applications across various technical domains."""
}
for filepath,content in sample_texts.items():
    with open(filepath,"w",encoding="utf-8") as f:
        f.write(content)
print("Text file created")
```

```python
from langchain_community.document_loaders import TextLoader
loader  = TextLoader("../../data/text_files/python_intro.txt", encoding="utf-8")
document = loader.load()
print(document)
```

```python
from langchain_community.document_loaders import DirectoryLoader
dir_loader = DirectoryLoader(
    "../../data/text_files",
    glob="**/*.txt",## Pattern to match files
    loader_cls=TextLoader, ## loader class to use
    loader_kwargs={"encoding": "utf-8"},
    show_progress=False
    )
documents = dir_loader.load()
print(documents)
```

```python
from langchain_community.document_loaders import PyPDFLoader,PyMuPDFLoader
dir_loader = DirectoryLoader(
    "../../data/pdf",
    glob="**/*.pdf",## Pattern to match files
    loader_cls=PyMuPDFLoader, ## loader class to use
    show_progress=False
    )
pdf_documents = dir_loader.load()
print(pdf_documents)
```

```python
from langchain_community.document_loaders import CSVLoader
dir_loader = DirectoryLoader(
    "../../data/csv",
    loader_cls=CSVLoader,
    loader_kwargs={"encoding": "utf-8"},
)
csv_documents = dir_loader.load()
print(csv_documents)
```

```python
from langchain_community.document_loaders import UnstructuredExcelLoader
dir_loader = DirectoryLoader(
    "../../data/excel",
    loader_cls=UnstructuredExcelLoader,
)
excel_documents = dir_loader.load()
print(excel_documents)
```
